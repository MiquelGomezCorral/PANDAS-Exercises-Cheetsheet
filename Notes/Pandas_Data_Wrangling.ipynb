# Dates

```python
data = {
    "date": ["2023-01-01", "2023-02-01", "2023-03-01"],
    "value": [10, 20, 30]
}
df = pd.DataFrame(data)

# Convert 'date' column to datetime
df["date"] = pd.to_datetime(df["date"])
```

### Extract Date components

```python
df["year"] = df["date"].dt.year
df["month"] = df["date"].dt.month
df["day"] = df["date"].dt.day
df["day_of_week"] = df["date"].dt.dayofweek  # Monday=0, Sunday=6
df["day_name"] = df["date"].dt.day_name()  # Full day name
```

### Filter by date and sample

```python
filtered_df = df[(df["date"] >= "2023-01-01") & (df["date"] <= "2023-02-28")]

# Resample by month and calculate the mean OF ALL THE VALUES IN TAHT MONTH. 
# Groups the data by month and applies an aggregation function
monthly_df = df.resample("M").mean()

# Convert datetime to period: from 2023-01-02 to 2023-01
df["month_period"] = df["date"].dt.to_period("M")

# Convert period back to datetime: from 2023-01 becomes 2023-01-01 00:00:00.
df["month_start"] = df["month_period"].dt.start_time
```

### Change / shift and Differences

```python
# Shift dates forward by 1 day
df["date_shifted"] = df["date"] + pd.Timedelta(days=1)

# Shift dates backward by 1 week
df["date_shifted"] = df["date"] - pd.Timedelta(weeks=1)


# Differences
df["days_since_start"] = (df["date"] - df["date"].min()).dt.days
```

### Format and timezones

```python
df["date_formatted"] = df["date"].dt.strftime("%Y-%m-%d")

# Localize to a time zone
df["date"] = df["date"].dt.tz_localize("UTC")

# Convert to another time zone
df["date"] = df["date"].dt.tz_convert("US/Eastern")
```
